```python
import numpy as np

def sugeno_fuzzy(suhu, trombosit, leukosit):

    # membership sederhana (sama dengan Mamdani)
    mu_suhu_tinggi = max(0, min((suhu-37.5)/1, 1))
    mu_suhu_demam = max(0, min((suhu-37)/1, (38.5-suhu)/1))

    mu_trombosit_rendah = max(0, min((150-trombosit)/50, 1))
    mu_trombosit_normal = max(0, min((trombosit-120)/100, 1))

    mu_lekopenia = max(0, min((4.5-leukosit)/2, 1))
    mu_lek_normal = max(0, min((leukosit-3.5)/3, 1))

    # RULE firing strength
    r1 = min(mu_suhu_tinggi, mu_trombosit_rendah, mu_lekopenia)
    r2 = min(mu_suhu_demam, mu_trombosit_rendah)
    r3 = min(mu_suhu_demam, mu_trombosit_normal)
    r4 = min(mu_suhu_tinggi, mu_lek_normal)

    # Sugeno output (konstanta)
    z1 = 8
    z2 = 8
    z3 = 5
    z4 = 5

    num = r1*z1 + r2*z2 + r3*z3 + r4*z4
    den = r1 + r2 + r3 + r4 + 1e-6

    return num/den

# uji data contoh
hasil = sugeno_fuzzy(38.5, 120, 3.8)
print("Sugeno risk score =", hasil)
```
